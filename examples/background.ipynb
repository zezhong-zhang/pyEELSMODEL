```python
import numpy as np
import matplotlib.pyplot as plt

# Define the curves and the blending function
def y1(x):
    return x**2

def y2(x):
    return np.sin(x)

def blending_function(x, x0, epsilon):
    t = (x - (x0 - epsilon)) / (2 * epsilon)
    t = np.clip(t, 0, 1)  # Ensure t is within [0, 1]
    return 10*t**3 - 15*t**4 + 6*t**5

# Define the combined curve
def combined_curve(x, x0, epsilon):
    B = blending_function(x, x0, epsilon)
    return (1 - B) * y1(x) + B * y2(x)

# Define the interval and transition point
x0 = 1.0
epsilon = 0.5
x = np.linspace(0, 2, 400)
y = combined_curve(x, x0, epsilon)

# Plot the curves
plt.plot(x, y1(x), label='$y_1(x) = x^2$', linestyle='dashed')
plt.plot(x, y2(x), label='$y_2(x) = \sin(x)$', linestyle='dashed')
plt.plot(x, y, label='Combined curve', linewidth=2)
plt.axvline(x=x0, color='gray', linestyle='dotted', label='$x_0$')
plt.legend()
plt.xlabel('$x$')
plt.ylabel('$y$')
plt.title('Smooth Combination of Two Curves')
plt.show()
```
